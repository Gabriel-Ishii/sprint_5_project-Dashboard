Introdução

Neste projeto, será realizada uma análise exploratória de um conjunto de dados contendo anúncios de veículos usados. O objetivo é identificar padrões e relações entre as principais características dos veículos e seus preços.

Durante a análise, serão investigadas variáveis como marca, ano de fabricação, quilometragem e preço dos veículos. Também serão tratados valores ausentes e possíveis valores discrepantes que possam distorcer os resultados.

A partir dos dados preparados, serão construídas visualizações para compreender melhor a distribuição dos preços, comparar os preços entre diferentes marcas e anos de fabricação e analisar a relação entre quilometragem e preço.

Os resultados desta análise servirão de base para a construção de um dashboard interativo, desenvolvido posteriormente com Streamlit.

```python
# Importando bibliotecas

import pandas as pd
import plotly.express as px
import streamlit as st

# Ajustando as configurações de exibição do pandas

pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
```

```python
# Carregando o dataset de veículos

vehicles = pd.read_csv(r'C:\sprint5dir\sprint_5_project-Dashboard\vehicles.csv')

# Exibindo informações sobre o dataset

vehicles.info()
print()
vehicles.head(10)
```

```python
# Verificando duplicatas e valores nulos no dataset

print(vehicles.duplicated().sum())
print()
print(vehicles.isnull().sum())
```

```python
# Analisando estatísticas descritivas dos preços dos carros do dataset

print(vehicles['price'].min(), vehicles['price'].max())
```

```python
# Analisando a quantidade de veículos com preço igual a 1 e exibindo os 15 veículos mais baratos

print((vehicles['price'] == 1).sum())
print()
vehicles.sort_values(by='price').head(15)
```

```python
# Analisando os veículos mais caros do dataset e exibindo os 17 veículos mais caros

vehicles.sort_values(by='price', ascending=False).head(17)
```

```python
# Removendo duplicatas e filtrando veículos com preço entre 500 e 110000
# Essa filtragem é feita para remover valores extremos, que pela analise exploratória, não tem valores coerentes com o mercado de veículos, e que podem prejudicar a análise dos dados.

vehicles = vehicles.drop_duplicates()
vehicles = vehicles[vehicles['price'].between(500, 110000)].reset_index(drop=True)
vehicles.info()
```

```python
# Criando uma nova coluna 'brand' a partir da coluna 'model', extraindo a primeira palavra de cada modelo como a marca do veículo

vehicles['brand'] = vehicles['model'].str.split().str[0]
vehicles['brand'].value_counts()
```

```python
# Analisando a médiana de preço dos veículos por marca e exibindo

vehicles_by_brand = vehicles.groupby('brand')
vehicles_by_brand_median = vehicles_by_brand['price'].median().round(2).sort_values(ascending=False)
print(vehicles_by_brand_median.head(20))
```

```python
# Construindo gráfico de preço mediano por marca

brands = vehicles_by_brand_median.reset_index()
brands = brands.sort_values(by='price')
palette = px.colors.sample_colorscale(
    'Blues',
    [
        0.25 + i * (0.75 / (len(brands) - 1))
        for i in range(len(brands))
    ]
)

fig_brand_bars = px.bar(
    brands,
    x='brand',
    y='price',
    color='brand',
    color_discrete_sequence=palette,
    title='Preço mediano por marca',
    labels={
        'brand': 'Marca',
        'price': 'Preço mediano ($)'
    }
)

fig_brand_bars.update_layout(
    height=600,
    legend_title_text='Marca'
)

fig_brand_bars.show()
```

```python
# Removendo veículos com odômetro nulo e analisando a verificando a quantidade de valores nulos na coluna 'odometer' após a remoção

vehicles_odometer = vehicles.dropna(subset=['odometer'])
print(vehicles_odometer.isna().sum())
```

```python
# Construindo gráfico de dispersão entre odômetro e preço

odometer = vehicles_odometer.reset_index(drop=True)

fig_odometer = px.scatter(
    odometer,
    x='odometer',
    y='price',
    title='Relação entre quilometragem e preço',
    labels={
        'odometer': 'Odômetro (milhas)',
        'price': 'Preço ($)'
    },
    opacity=0.18,
    render_mode='webgl',
    log_y=True
)

fig_odometer.update_traces(
    marker=dict(size=4)
)

fig_odometer.update_layout(
    height=650,
    template='plotly_white',
    title_x=0.5,
    xaxis=dict(
        showgrid=True,
        gridcolor='lightgray'
    ),
    yaxis=dict(
        showgrid=True,
        gridcolor='lightgray',
        tickprefix='$'
    )
)

fig_odometer.show()
```

```python
# Analisando a quantidade de valores nulos na coluna 'model_year' e exibindo o menor e maior ano de fabricação dos veículos

print(vehicles['model_year'].min(), vehicles['model_year'].max())
vehicles['model_year'].isna().sum()

# Removendo veículos com ano de fabricação nulo e filtrando marcas com mais de 1000 veículos no dataset

vehicles_filtered_year = vehicles.dropna(subset=['model_year']).copy()

brand_over_1000 = vehicles_filtered_year['brand'].value_counts()
brand_over_1000 = brand_over_1000[brand_over_1000 >= 1000]
print(brand_over_1000)

# Agrupando os veículos por marca e ano de fabricação, calculando a mediana de preço para cada grupo e exibindo o resultado

vehicles_filtered_year = vehicles_filtered_year[vehicles_filtered_year['brand'].isin(brand_over_1000.index)]
vehicles_grouped_year = vehicles_filtered_year.groupby(['brand', 'model_year'])['price'].median().reset_index()
print(vehicles_grouped_year)
```

```python
fig_year_bars = px.bar(
    vehicles_grouped_year,
    x='brand',
    y='price',
    barmode='group',
    title='Preço Mediano por Marca e Ano de Fabricação',
    color='model_year',
    color_discrete_sequence=px.colors.sequential.Blues,
    labels={
        'price': 'Preço ($)',
        'brand': 'Marca',
        'model_year': 'Ano de Fabricação'
    },

)

fig_year_bars.update_layout(
    height=800,
    title_x=0.5,
    legend_title_text='Ano de Fabricação',
    yaxis_title='Preço Mediano ($)',
)

fig_year_bars.show()
```
